```python
def standardize_time_and_resample(df, interval="1s"):
    df = df.copy()
    df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S.%f")
    df = df.set_index("Time").sort_index()

    r = df.resample(interval).mean()

    r = r.interpolate(method="time", limit_direction="both")

    return r.reset_index()

standardized_n_dfs = [standardize_time_and_resample(df) for df in ordered_n_dfs]
standardized_p_dfs = [standardize_time_and_resample(df) for df in ordered_p_dfs]
```

```python
def extract_features(df_list, period=12):
    all_features = []
    processed_dfs = []
    t_win = 121 
    
    for df in df_list:
        temp = df.copy()
        
        # 1. STL Trend
        temp['total_dp_trend'] = STL(temp['Total_System_DP'], period=period, trend=t_win, robust=True).fit().trend
        temp['p_balance'] = temp['TS_DP'] / (temp['Pump_to_TS_DP'] + 1e-6)
        temp['p_balance_trend'] = STL(temp['p_balance'], period=period, trend=t_win, robust=True).fit().trend
        
        # A. Relative Change (Normalization)
        # We divide by the median of the first 30 seconds of THIS run
        base_dp = temp['total_dp_trend'].iloc[:30].median()
        temp['rel_dp_change'] = temp['total_dp_trend'] / base_dp
        
        # B. The Gradient (Slope)
        # Is the pressure accelerating? (Change over 10 seconds)
        temp['dp_slope'] = temp['total_dp_trend'].diff(periods=10)
        
        # 3. Volatility
        resids = STL(temp['Total_System_DP'], period=period, trend=t_win, robust=True).fit().resid
        temp['volatility'] = resids.rolling(window=12).std()
        
        temp = temp.dropna()
        processed_dfs.append(temp)
        
        # Use only the relative/dynamic features
        features = ['rel_dp_change', 'p_balance_trend', 'volatility', 'dp_slope']
        all_features.append(temp[features])
        
    return pd.concat(all_features), processed_dfs
```

```python
X_train_raw, clean_processed = extract_features(standardized_n_dfs)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
```

```python
model = IsolationForest(
    n_estimators=150, 
    contamination=0.005,
)
model.fit(X_train_scaled)
```

```python
# 3. Extract features from PLUG data
_, plug_processed = extract_features(standardized_p_dfs)

# 4. Score every file
scored_plug_dfs = []
features_list = ['rel_dp_change', 'p_balance_trend', 'volatility', 'dp_slope']

for df in plug_processed:
    temp = df.copy()
    X_test = scaler.transform(temp[features_list])
    
    # decision_function: Positive = Normal, Negative = Anomaly
    temp['anomaly_score'] = model.decision_function(X_test)
    
    # Smooth the score to avoid 'flickering' alarms
    temp['smooth_score'] = temp['anomaly_score'].rolling(window=6).mean()
    scored_plug_dfs.append(temp)
```

```python
# 1. Score the Clean (Normal) Files
scored_clean_dfs = []
for df in clean_processed:
    temp = df.copy()
    X_test = scaler.transform(temp[features_list])
    temp['anomaly_score'] = model.decision_function(X_test)
    temp['smooth_score'] = temp['anomaly_score'].rolling(window=6).mean()
    scored_clean_dfs.append(temp)

# 2. Now calculate the Minima safely
# We use dropna() because the rolling window creates NaNs at the start
min_n = min([df['smooth_score'].dropna().min() for df in scored_clean_dfs])
min_p = min([df['smooth_score'].dropna().min() for df in scored_plug_dfs])

print(f"Deepest Anomaly in Normal Runs: {min_n:.4f}")
print(f"Deepest Anomaly in Plug Runs:   {min_p:.4f}")
```

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import RobustScaler
from sklearn.decomposition import PCA
from sklearn.ensemble import IsolationForest
import matplotlib.pyplot as plt

USE_COLS = [
    'TS outlet pressure (Mean)',
    'TS inlet pressure (Mean)',
    'Pump outlet pressure (Mean)',
    'Temperature TS outlet (Mean)',
    'Tank temperature (Mean)',
    'Temperature TS inlet (Mean)',
    'Bypass temperature (Mean)',
    'TS_DP',
    'Pump_to_TS_DP',
    'Total_System_DP'
]
```

```python
def add_extra_features(df):
    df = df.copy()
    eps = 1e-6
    df["Pump_to_TS_over_TS"] = df["Pump_to_TS_DP"] / (df["TS_DP"] + eps)
    df["TS_over_Total"] = df["TS_DP"] / (df["Total_System_DP"] + eps)
    return df
```

```python
def make_window_features(df, base_cols, window="60s"):
    """
    Returns a dataframe indexed by Time with window features for unsupervised learning.
    window: e.g. "60s", "120s", "300s"
    """
    df = df.copy()
    if "Time" in df.columns:
        df = df.set_index("Time")
    df = df.sort_index()
    
    # ensure numeric
    X = df[base_cols].astype("float64")
    
    roll = X.rolling(window, min_periods=10)

    feats = pd.DataFrame(index=X.index)

    # basic stats
    for c in base_cols:
        feats[f"{c}__mean"] = roll[c].mean()
        feats[f"{c}__std"]  = roll[c].std()

        # robust MAD
        feats[f"{c}__mad"]  = roll[c].apply(lambda x: np.median(np.abs(x - np.median(x))), raw=True)

        # spike rate: fraction of samples in window beyond 6*MAD from rolling median
        med = roll[c].median()
        mad = feats[f"{c}__mad"] + 1e-6
        z = (X[c] - med) / mad
        feats[f"{c}__spike_rate"] = (np.abs(z) > 6).rolling(window, min_periods=10).mean()

        # slope: (value - rolling mean at start) approx via linear fit is expensive;
        # cheap proxy: rolling mean of first derivative
        dt = X.index.to_series().diff().dt.total_seconds().replace(0, np.nan)
        d1 = X[c].diff() / dt
        feats[f"{c}__slope"] = d1.rolling(window, min_periods=10).mean()

    feats = feats.replace([np.inf, -np.inf], np.nan).dropna()
    return feats.reset_index()
```

```python
WINDOW = "60s"  # try 60s, 120s, 300s depending on how fast things change

train_feats_list = []
for df in standardized_n_dfs:
    df2 = add_extra_features(df)
    feats = make_window_features(df2, USE_COLS + ["Pump_to_TS_over_TS", "TS_over_Total"], window=WINDOW)
    train_feats_list.append(feats)

train_feats = pd.concat(train_feats_list, ignore_index=True)
train_time = train_feats["Time"]
X_train = train_feats.drop(columns=["Time"])
```

```python
scaler = RobustScaler()
Xtr = scaler.fit_transform(X_train)

# keep enough components to explain most variance
pca = PCA(n_components=0.98, random_state=42)
Ztr = pca.fit_transform(Xtr)
Xtr_hat = pca.inverse_transform(Ztr)

train_rec_err = np.mean((Xtr - Xtr_hat)**2, axis=1)
```

```python
iso = IsolationForest(
    n_estimators=400,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)
iso.fit(Xtr)
train_if_score = -iso.score_samples(Xtr)  # higher = more anomalous
```

```python
PCA_THRESH = np.quantile(train_rec_err, 0.995)
IF_THRESH  = np.quantile(train_if_score, 0.995)

PCA_THRESH, IF_THRESH
```

```python
def score_run(df, window=WINDOW):
    df2 = add_extra_features(df)
    feats = make_window_features(df2, USE_COLS + ["Pump_to_TS_over_TS", "TS_over_Total"], window=window)
    times = feats["Time"]
    X = feats.drop(columns=["Time"])
    Xs = scaler.transform(X)

    # PCA reconstruction error
    Z = pca.transform(Xs)
    X_hat = pca.inverse_transform(Z)
    rec_err = np.mean((Xs - X_hat)**2, axis=1)

    # Isolation Forest score
    if_score = -iso.score_samples(Xs)

    out = pd.DataFrame({
        "Time": times,
        "pca_rec_err": rec_err,
        "if_score": if_score
    })
    return out
```

```python
plug_scores = [score_run(df) for df in standardized_p_dfs]
no_plug_scores = [score_run(df) for df in standardized_n_dfs]
```

```python
def plot_scores(score_df, title=""):
    s = score_df.set_index("Time")
    plt.figure()
    s["pca_rec_err"].rolling("30s").mean().plot()
    plt.axhline(PCA_THRESH, linestyle="--")
    plt.title(f"{title} PCA reconstruction error (smoothed)")
    plt.show()

    plt.figure()
    s["if_score"].rolling("30s").mean().plot()
    plt.axhline(IF_THRESH, linestyle="--")
    plt.title(f"{title} IsolationForest score (smoothed)")
    plt.show()
```

```python
def make_alarm(score_df, score_col, thresh, sustain="30s", frac=0.6):
    s = score_df.set_index("Time")[score_col]
    above = (s > thresh).astype(float)
    alarm_frac = above.rolling(sustain).mean()
    alarm = alarm_frac > frac
    return pd.DataFrame({"score": s, "alarm_frac": alarm_frac, "alarm": alarm})

alarm_df = make_alarm(plug_scores[0], "pca_rec_err", PCA_THRESH, sustain="30s", frac=0.6)

plt.figure()
alarm_df["alarm_frac"].plot()
plt.axhline(0.6)
plt.title("Alarm fraction (PCA)")
plt.show()
```
